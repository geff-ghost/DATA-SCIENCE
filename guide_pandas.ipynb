# How to calculate summary statistics
# Aggregating statistics
what is the average of the Titanic passengers?

```python
titanic = pd.read_csv("data/titanic.csv")
titanic.head()
```

```python
titanic["Age"].mean()
```

```python
titanic["Age"].min()
```

What is the median age and ticket fare price of the Titanic passengers?

```python
titanic[["Age", "Fare"]].median()
```

```python
titanic[["Age", "Fare"]].describe()
```

```python
titanic.agg(
    {
        "Age": ["min", "max", "median", "skew"],
        "Fare": ["min", "max", "median", "mean"]
    }
)
```

# Aggregating statistics grouped by category
What is the average age for male versus femal Titanic passengers?

```python
titanic[["Sex", "Age"]].groupby("Sex").mean()
```

```python
titanic[["Sex", "Age"]].groupby("Sex").sum()
```

```python
titanic.groupby("Sex").mean(numeric_only=True)
```

```python
titanic.groupby("Sex")["Age"].mean()
```

What is the mean ticket fare price for each of the sex and cabin class combinations?

```python
titanic.groupby(["Sex", "Pclass"])["Fare"].mean()
```

```python
titanic.groupby("Pclass")["Pclass"].mean()
```

# Count number of records by category
What is the number of passengers in each of the cabin classes?

```python
titanic["Pclass"].value_counts()
```

```python
titanic.groupby("Pclass")["Sex"].count()
```

```python
titanic.groupby(["Pclass", "Sex"])["Sex"].count()
```

```python
titanic.groupby("Pclass")["Pclass"].count()
```
